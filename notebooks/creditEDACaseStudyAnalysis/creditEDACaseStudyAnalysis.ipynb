### Handilling missing values

##### Finding the missing values in the dataset

```python
df_app.isnull().values.any()
```

```python
df_app.isnull().values.sum()
```

```python
# Finding the columns with null values
df_app.columns[df_app.isnull().any()]
```

```python
#Finding the number of columns with null values
len(df_app.columns[df_app.isnull().any()])
```

```python
# Finding the null count and percentage
null_count = df_app.isnull().sum()
null_percentage = df_app.isnull().sum()/len(df_app)*100
```

```python
# Creating null dataframe
null_df = pd.DataFrame({'Column_Name': df_app.columns, 'Null_values': null_count, 'Null_Percentage': null_percentage})
null_df.reset_index(drop=True, inplace=True)
```

```python
null_df.sort_values('Null_Percentage', ascending=False)
```

#### Finding columns with missing values more than 40%

```python
# Finding the names of the columns with more than 40% Null Values
null_df[null_df['Null_Percentage']>40]['Column_Name']
```
